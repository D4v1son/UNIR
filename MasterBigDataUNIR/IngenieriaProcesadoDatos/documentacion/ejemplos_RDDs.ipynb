# Ejemplo de uso de RDDs

* La SparkSession ya viene creada en la variable spark 
* Utilizamos el SparkContext que hay dentro de la SparkSession. El SparkContext era en versiones primitivas de Spark el equivalente a la SparkSession actual

```python
flights_lines = spark.sparkContext.textFile("gs://unirbucket2023/datos/flights.csv").cache()
```

```python
conteo = flights_lines.count()
```

```python
lista_cinco = flights_lines.take(5) # esto provoca que se tenga que materializar flights_lines
```

```python
lista_cinco
```

```python
def trocear(linea):
    return linea.split(",")

def no_empieza_por_year(linea):
    return not linea.startswith("year")

flights_tuples_rdd = flights_lines\
                        .filter(lambda cadena: not cadena.startswith("year"))\
                        .map(trocear)\
                        .filter(lambda lista: lista[4] != "NA" and lista[5] != "NA")\
                        .map(lambda lista: (lista[10], lista[11], int(lista[4]), int(lista[5])))\
                        .filter(lambda tupla: tupla[0] == "SEA")
```

```python
lista_tuplas = flights_tuples_rdd.take(3)
```
