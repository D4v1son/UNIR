```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
```

```python
# La base de datos que vamos a utilizar ya se encuentra dentro de las librerías

dfs = sns.load_dataset("mpg").dropna()

x = dfs["horsepower"]

y = dfs["mpg"]

print("Correlación: ", np.corrcoef(x,y)[0,1])

X = sm.add_constant(x)
model = sm.OLS(y, X).fit()
res = model.resid

plt.scatter(x,y)
plt.plot(x, model.predict(X))
plt.title("Positiva fuerte (horsepower vs mpg)")
plt.show()

plt.scatter(x,res)
plt.axhline(0)
plt.title("Residuos: aleatorios (buen ajuste)")
plt.show()
```
