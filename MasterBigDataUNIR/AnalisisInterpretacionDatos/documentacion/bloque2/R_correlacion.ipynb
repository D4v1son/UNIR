```r
### peso del coche vs consumo

data(mtcars)

x <- mtcars$wt
y <- mtcars$mpg

cov(x, y)
cor(x, y)

modelo <- lm(mpg ~ wt, data = mtcars)

plot(x, y,
     main = "Correlación negativa fuerte",
     xlab = "Peso (wt)",
     ylab = "Consumo (mpg)")
abline(modelo, col = "blue")

plot(x, resid(modelo),
     main = "Residuos (buen ajuste)",
     xlab = "Peso",
     ylab = "Residuos")
abline(h = 0)
```

Interpretación

Correlación ≈ –0.87 (muy fuerte)

Residuos aleatorios → modelo adecuado

#### CASO 2: CORRELACIÓN POSITIVA MODERADA
Dataset: cars

Velocidad vs distancia de frenado

```r
data(cars)

x <- cars$speed
y <- cars$dist

cov(x, y)
cor(x, y)

modelo <- lm(dist ~ speed, data = cars)

plot(x, y,
     main = "Correlación positiva moderada",
     xlab = "Velocidad",
     ylab = "Distancia")
abline(modelo, col = "blue")

plot(x, resid(modelo),
     main = "Residuos con patrón (no lineal)",
     xlab = "Velocidad",
     ylab = "Residuos")
abline(h = 0)
```

Clave didáctica

Correlación alta

Residuos NO aleatorios

❌ Modelo lineal incorrecto

#### CASO 3: CORRELACIÓN POSITIVA FUERTE

#### CASO 5: SIN CORRELACIÓN (PRÁCTICAMENTE NULA)
Dataset: faithful

Tiempo de espera vs duración (ejemplo para contraste)

```r
data(faithful)

x <- faithful$waiting
y <- faithful$eruptions

cov(x, y)
cor(x, y)

modelo <- lm(eruptions ~ waiting, data = faithful)

plot(x, y,
     main = "Relación aparente (mixta)",
     xlab = "Tiempo de espera",
     ylab = "Duración")
abline(modelo, col = "blue")

plot(x, resid(modelo),
     main = "Residuos con dos grupos",
     xlab = "Tiempo de espera",
     ylab = "Residuos")
abline(h = 0)
```

Clave

Grupos ocultos

Importancia de segmentar datos
