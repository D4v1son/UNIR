```python
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# parametros poblacionales
np.random.seed(42)
population_size = 100000
population = np.random.normal(scale=2, size=population_size) # poblacion no normal, distribución exponencial
```

```python
# configurar la simulacion
sample_size = [5, 30, 100] # tamaños de muestra a analizar
num_samples = 1000 # numero de muestras a generar para cada tamaño de muestra
```

```python
plt.figure(figsize=(12, 8))

for i, sample_size in enumerate(sample_size):
    sample_means = []
    
    # generar muestras y calcular sus medias
    for j in range(num_samples):
        sample = np.random.choice(population, size=sample_size, replace=False)
        sample_means.append(np.mean(sample))
        
    # generar grafico de la distribución de las medias muestrales
    sns.histplot(sample_means, kde=True, stat="density", bins=30, label=f'n={sample_size}')
    
plt.title('Distribución de las Medias Muestrales para Diferentes Tamaños de Muestra', fontsize=14)
plt.xlabel('Media Muestral', fontsize=12)
plt.ylabel('Densidad', fontsize=12)
plt.legend(title='Tamaño de Muestra')
plt.show()
```
