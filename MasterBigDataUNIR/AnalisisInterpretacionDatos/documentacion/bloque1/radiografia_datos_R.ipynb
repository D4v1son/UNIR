```r
# ============================================================
# 1. CARGA DE LIBRERÍAS
# ============================================================
# ggplot2 → gráficos
# dplyr   → manipulación de datos
library(ggplot2)
library(dplyr)

# ============================================================
# 2. LECTURA DEL CSV
# ============================================================
# El archivo está separado con ';' y tiene caracteres latinos (acentos)
df <- read.csv("calidad_aire_estaciones.csv",
               sep = ";",
               fileEncoding = "latin1",
               quote = "\"",
               fill = TRUE)

# ============================================================
# 3. INFORMACIÓN GENERAL DEL DATASET
# ============================================================
cat("\n=== INFORMACIÓN GENERAL ===\n")
str(df)   # tipos de datos

cat("\n=== VALORES FALTANTES POR COLUMNA ===\n")
print(colSums(is.na(df)))   # cantidad de NA por columna

cat("\n=== ESTADÍSTICAS BÁSICAS ===\n")
print(summary(df))          # estadísticas descriptivas


# ============================================================
# 4. RADIOGRAFÍA DE CADA COLUMNA
# ============================================================
for (col in names(df)) {

  cat("\n=====================================\n")
  cat("RADIODRAFÍA DE LA COLUMNA:", col, "\n")
  cat("=====================================\n\n")

  # Tipo de dato
  print(paste("Tipo:", class(df[[col]])))

  # Valores faltantes
  print(paste("NA:", sum(is.na(df[[col]]))))

  # Estadísticas
  print(summary(df[[col]]))

  # ----------------------------------------------------------
  # Si NO es numérica, omitimos gráficos
  # ----------------------------------------------------------
  if (!is.numeric(df[[col]])) {
    cat("(Columna NO numérica: no se grafican histogramas ni boxplots)\n")
    next
  }

  # ----------------------------------------------------------
  # HISTOGRAMA: distribución de la variable
  # ----------------------------------------------------------
  g1 <- ggplot(df, aes(x = !!sym(col))) +
    geom_histogram(bins = 20) +
    ggtitle(paste("Histograma de", col))

  print(g1)

  # ----------------------------------------------------------
  # BOXPLOT: detectar la mediana y outliers
  # ----------------------------------------------------------
  g2 <- ggplot(df, aes(x = "", y = !!sym(col))) +
    geom_boxplot() +
    ggtitle(paste("Boxplot de", col))

  print(g2)

  # ----------------------------------------------------------
  # CONCLUSIONES PARA LOS ALUMNOS
  # ----------------------------------------------------------
  cat("\n🔎 CONCLUSIONES PEDAGÓGICAS:\n")
  cat("
1) HISTOGRAMA:
   - Muestra la forma de la distribución
   - Permite ver si hay sesgo (asimetría)
   - Detecta rangos frecuentes y valores raros

2) BOXPLOT:
   - Permite ver mediana, cuartiles y dispersión
   - Identifica valores atípicos (outliers)
   - Muestra si la variable es simétrica o sesgada
\n")
}

cat("\n=== ANÁLISIS COMPLETO FINALIZADO ===\n")
```
