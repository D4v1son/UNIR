```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ============================================================
# 1. LECTURA DEL CSV
# ============================================================
# El archivo tiene separador con ';' y codificación Latin1 (por ser texto en español)
# También tiene comillas irregulares, por eso usamos engine="python"
df = pd.read_csv(
    "calidad_aire_estaciones.csv",
    sep=";",
    encoding="latin1",
    quotechar='"',
    engine="python",
    on_bad_lines="skip"   # Evita errores por líneas mal formadas
)

# ============================================================
# 2. INFORMACIÓN GENERAL DEL DATASET
# ============================================================
print("\n=== INFORMACIÓN GENERAL DEL DATASET ===\n")
print(df.info())  # Muestra tipos de datos y columnas

print("\n=== ESTADÍSTICAS BÁSICAS ===\n")
print(df.describe(include="all"))  # Estadísticas numéricas y categóricas

print("\n=== VALORES FALTANTES POR COLUMNA ===\n")
print(df.isna().sum())  # Conteo de valores nulos por columna


# ============================================================
# 3. RADIOGRAFÍA SISTEMÁTICA DE CADA COLUMNA
# ============================================================
for col in df.columns:

    print("\n========================================")
    print(f"RADIODRAFÍA DE LA COLUMNA: {col}")
    print("========================================\n")

    # Tipo de dato
    print(f"Tipo de dato: {df[col].dtype}")

    # Cantidad de valores faltantes
    print(f"Valores faltantes: {df[col].isna().sum()}")

    # Estadísticas básicas
    print("\nEstadísticas descriptivas:")
    print(df[col].describe())

    # ¿Es columna numérica? → Si lo es, hacemos gráficos
    if pd.api.types.is_numeric_dtype(df[col]):

        # Crear figura con histograma + boxplot
        plt.figure(figsize=(12,4))

        # --------------------------------------------------------
        # HISTOGRAMA (muestra la distribución)
        # --------------------------------------------------------
        plt.subplot(1,2,1)
        plt.hist(df[col].dropna(), bins=20)
        plt.title(f"Histograma de {col}")
        plt.xlabel(col)
        plt.ylabel("Frecuencia")

        # --------------------------------------------------------
        # BOXPLOT (muestra medianas y valores atípicos)
        # --------------------------------------------------------
        plt.subplot(1,2,2)
        sns.boxplot(x=df[col], orient="h")
        plt.title(f"Boxplot de {col}")

        plt.tight_layout()
        plt.show()

        # --------------------------------------------------------
        # CONCLUSIONES PEDAGÓGICAS
        # --------------------------------------------------------
        print("\n🔎 CONCLUSIONES QUE LOS ALUMNOS DEBEN INTERPRETAR:")

        print("""
1) **Histograma:**
   - Permite ver si la distribución es:
     • uniforme  
     • sesgada a la izquierda o derecha  
     • normal o parecida  
     • multimodal (varios picos)
   - Ayuda a detectar rangos frecuentes y valores raros.

2) **Boxplot:**
   - Muestra:
     • mínimo y máximo sin atípicos  
     • cuartiles (Q1, Q2/mediana, Q3)  
     • outliers (puntos fuera de la caja)
   - Sirve para ver:
     • si la variable está centrada o desbalanceada  
     • si hay valores extremos que puedan afectar análisis futuros
""")

    else:
        print("(Columna NO numérica: no se generan gráficos)\n")

print("\n=== ANÁLISIS COMPLETO FINALIZADO ===")
```
