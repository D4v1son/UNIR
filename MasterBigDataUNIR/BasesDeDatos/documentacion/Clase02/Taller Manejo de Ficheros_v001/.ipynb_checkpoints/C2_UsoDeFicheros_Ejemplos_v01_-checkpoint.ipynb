## Ficheros CSV

```python
# Usando la biblioteca csv
import pandas as pd
import glob

# Obtener una lista de todos los archivos CSV en el directorio
csv_files = glob.glob('./data/*.csv')

# Crear una lista para almacenar los DataFrames
dataframes = []

# Leer cada archivo CSV y agregarlo a la lista
for csv in csv_files:
    #df = pd.read_csv(csv)
    df = pd.read_csv(csv, on_bad_lines='skip', delimiter=";")
    dataframes.append(df)

# Concatenar todos los DataFrames en uno 
# RETO, revisar los ID y crear consecutivos
combined_df = pd.concat(dataframes)
```

## Ficheros XML

```python
import xml.etree.ElementTree as ET

# Cargar el archivo XML
tree = ET.parse('./data/0000-0001-5000-5000.xml')

# Obtener la raíz del XML
root = tree.getroot()

# Imprimir cada elemento del XML
for child in root:
    print(child.tag, child.attrib)
```

```python
# Cargar el archivo XML
tree = ET.parse('./data/0000-0001-5000-5000.xml')

# Obtener la raíz del XML
root = tree.getroot()

# Crear un diccionario para almacenar la información
data = {}

# Recorrer cada elemento del XML
for child in root:
    # Si el elemento tiene atributos, agregarlos al diccionario
    if child.attrib:
        data[child.tag] = child.attrib
    # Si el elemento tiene texto, agregarlo al diccionario
    if child.text.strip():
        data[child.tag] = child.text.strip()

print(data)
```

```python
import xml.etree.ElementTree as ET

# Parsear el archivo XML
tree = ET.parse('./data/0000-0001-5000-5000.xml')
root = tree.getroot()

# Obtener el título del rol
if root.find('.//{http://www.orcid.org/ns/activities}role-title'):
    role_title = root.find('.//{http://www.orcid.org/ns/activities}role-title').text
    print('Título del rol:', role_title)

# Obtener la fecha de inicio y fin
start_date = root.find('.//{http://www.orcid.org/ns/common}start-date')
end_date = root.find('.//{http://www.orcid.org/ns/common}end-date')
print('Fecha de inicio:', start_date.find('{http://www.orcid.org/ns/common}year').text, start_date.find('{http://www.orcid.org/ns/common}month').text, start_date.find('{http://www.orcid.org/ns/common}day').text)
print('Fecha de fin:', end_date.find('{http://www.orcid.org/ns/common}year').text, end_date.find('{http://www.orcid.org/ns/common}month').text, end_date.find('{http://www.orcid.org/ns/common}day').text)

# Obtener el nombre de la organización
org_name = root.find('.//{http://www.orcid.org/ns/common}name').text
print('Nombre de la organización:', org_name)

# Obtener la ciudad y el país de la organización
org_city = root.find('.//{http://www.orcid.org/ns/common}city').text
org_country = root.find('.//{http://www.orcid.org/ns/common}country').text
print('Ciudad de la organización:', org_city)
print('País de la organización:', org_country)

# Obtener el identificador de la organización
org_id = root.find('.//{http://www.orcid.org/ns/common}disambiguated-organization-identifier').text
print('Identificador de la organización:', org_id)

# Obtener la URL
url = root.find('.//{http://www.orcid.org/ns/common}url').text
print('URL:', url)

# Obtener el título del trabajo
work_title = root.find('.//{http://www.orcid.org/ns/common}title').text
print('Título del trabajo:', work_title)

# Obtener la fecha de publicación del trabajo
pub_date = root.find('.//{http://www.orcid.org/ns/common}publication-date')
print('Fecha de publicación del trabajo:', pub_date.find('{http://www.orcid.org/ns/common}year').text, pub_date.find('{http://www.orcid.org/ns/common}month').text)

# Obtener el título del diario del trabajo
journal_title = root.find('.//{http://www.orcid.org/ns/work}journal-title').text
print('Título del diario del trabajo:', journal_title)
```
