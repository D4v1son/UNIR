## Obtención de datos históricos mediante API

[Documentación](https://steamapi.xpaw.me) no oficial pero muy útil de todas formas.

### Datos de mi api web
- Nombre del dominio: auge_del_indie

````python
import requests, pandas as pd

KEY = "TU_API_KEY"
apps, last = [], 0
while True:
    r = requests.get("https://api.steampowered.com/IStoreService/GetAppList/v1/",
                     params={"key": KEY, "max_results": 50000,
                             "include_games": "true", "last_appid": last})
    resp = r.json()["response"]
    apps += resp.get("apps", [])
    if not resp.get("have_more_results"):
        break
    last = resp["last_appid"]

df = pd.DataFrame(apps)
print(len(df))
df.to_csv("steam_apps.csv", index=False)
````
